```python
# Задача № 32
# Дано координаты векторов a и b
# Выход: 1. True если вектора коллинеарны, False в противном случае.
#        2. Число, показывающее во сколько раз отличается длина вектора а от длины вектора b.
#        3. True если вектора направлены противоположно, иначе False

a = np.array([3, 0, 0])
b = np.array([-5, 0, 0])

# ваш код

cross_a_b = np.cross(a, b)
if cross_a_b.mean() == cross_a_b[0]:
    res1 = True
else:
    res1 = False
l_a = np.linalg.norm(a)
l_b = np.linalg.norm(b)
res2 = l_a/l_b
if a.dot(b)/(l_a*l_b) == -1:
    res3 = True
else:
    res3 = False
print(res1)
print(res2)
print(res3)
```

```python
# Задача № 59
# Дано координаты вершин треугольника АВС
# Выход: длина АК - высота треугольника АВС

A = np.array([2, 2, 2])
B = np.array([4, 0, 3])
C = np.array([0, 1, 0])

# ваш код

ab = np.linalg.norm(B-A)
ac = np.linalg.norm(C-A)
bc = np.linalg.norm(C-B)
pp = 0.5*(ab+ac+bc)
s = (pp*(pp-ab)*(pp-ac)*(pp-bc))**0.5
h = 2*s / np.linalg.norm(bc)
print(h)
```
