```python
# Задача № 17
# Дано: два датасета с данными о температуре и влажности. Каждый датасет представлен в виде двумерного массива,
#       где строки представляют дни, а столбцы представляют конкретные моменты времени в течение дня.
#       Тип данных - int. Вам нужно изменить их на float.
# Выход: 1. Создайте функцию, которая принимает на вход два аргумента: dataset и new_dtype.
#        2. Используйте функцию для изменения типа данных каждого датасета на тип float.

# Датасет температуры (тип данных: int)
temp_data = np.array([[22, 25, 24, 23, 21],
                      [20, 18, 19, 22, 24]])

# Датасет влажности (тип данных: int)
humid_data = np.array([[45, 50, 48, 52, 47],
                       [45, 48, 50, 51, 53]])

# ваш код


def change_data_type(dataset, new_dtype):
    return dataset.astype(new_dtype) 

temp_data = change_data_type(temp_data, float)
humid_data = change_data_type(humid_data, float)

print(f'Новый тип данных temp_data: {temp_data.dtype}')
print(f'Новый тип данных humid_data: {humid_data.dtype}')
```

```python
# Задача № 128
# Выход: напишите функцию most_frequent_number(arr), которая находит наиболее
#        часто встречающееся число и количество его вхождений

arr = np.array([1, 2, 3, 1, 2, 1, 4, 5, 4, 3, 3])

# ваш код

def most_frequent_number(arr):
    bin_counts = np.bincount(arr)
    most_frequent_index = np.argmax(bin_counts)
    occurrences = bin_counts[most_frequent_index]
    return most_frequent_index, occurrences

result_number, result_occurrences = most_frequent_number(arr)

print(f"Наиболее часто встречающееся число: {result_number}")
print(f"Количество вхождений: {result_occurrences}")
```
